# pubify-mpl Quick Start

This notebook mirrors the minimal example from the README. It creates a simple Matplotlib figure, prepares the tracked example LaTeX project in `examples/tex/`, and exports the figure using the `onewide` layout.


````python
import matplotlib.pyplot as plt

from pubify_mpl import prepare, save_fig
````

````python
paper_dir = "tex"
figures_dir = "tex/figures"

template = {
    # Values for the default LaTeX article class example.
    "textwidth_in": 5.39643,
    "textheight_in": 7.58960,
    "base_fontsize_pt": 12.0,
    "caption_lineheight_pt": 13.6,
    "subcaption_lineheight_pt": 13.6,
}
````

````python
fig, ax = plt.subplots()
ax.plot([0, 1], [0, 1])
ax.set_xlabel("x")
ax.set_ylabel("y")
plt.show()
````

````python
prepare(paper_dir, template=template)
save_fig(fig, "onewide", f"{figures_dir}/plot.pdf", template=template)
````

After running the cells above, your LaTeX project will usually contain:

```text
tex/main.tex
tex/pubify.sty
tex/pubify-template.tex
tex/figures/plot.pdf
```

This notebook writes the generated files into the same `tex/` directory.

Then in `tex/main.tex` you can use:

```tex
\usepackage{pubify}
```
